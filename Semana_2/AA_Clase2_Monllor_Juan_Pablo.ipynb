## Ejercicio 1 - Adquisición de datos con Pandas

En este ejercicio se trabajará con datos de una empresa ficticia de comercio electrónico, utilizando diferentes formatos de archivos: CSV, JSON y Excel. Los datos serán importados y analizados mediante la librería Pandas.

### 1. Generación del archivo de ventas

Se genera un conjunto de datos ficticio con información de ventas de productos y se almacena en formato CSV.

```python
import pandas as pd

# Datos ficticios de ventas
ventas = {
    "id_venta": ["V001", "V002", "V003", "V004", "V005",
                 "V006", "V007", "V008", "V009", "V010"],
    "id_cliente": ["C001", "C002", "C003", "C001", "C004",
                   "C005", "C002", "C006", "C007", "C003"],
    "id_producto": ["P001", "P003", "P002", "P005", "P004",
                    "P001", "P006", "P003", "P007", "P002"],
    "cantidad": [1, 2, 1, 3, 2, 1, 1, 2, 1, 2],
    "precio_unitario": [850000, 25000, 450000, 18000, 65000,
                        850000, 120000, 25000, 95000, 450000],
    "fecha": ["2026-08-01", "2026-08-02", "2026-08-03", "2026-08-04",
              "2026-08-05", "2026-08-06", "2026-08-07", "2026-08-08",
              "2026-08-09", "2026-08-10"]
}

# Crear un DataFrame
df_ventas = pd.DataFrame(ventas)

# Guardar los datos en un archivo CSV
df_ventas.to_csv("ventas.csv", index=False)

# Mostrar los datos
df_ventas
```

### 5. Generación del archivo de inventario

Se genera un conjunto de datos ficticio con información sobre los productos y su stock disponible, y se almacena en formato Excel.

```python
# Datos ficticios del inventario
inventario = {
    "id_producto": ["P001", "P002", "P003", "P004",
                    "P005", "P006", "P007", "P008"],
    "producto": ["Notebook", "Smartphone", "Mouse", "Teclado",
                 "Auriculares", "Monitor", "Webcam", "Tablet"],
    "categoria": ["Computación", "Telefonía", "Accesorios", "Accesorios",
                  "Accesorios", "Computación", "Accesorios", "Telefonía"],
    "precio": [850000, 450000, 25000, 65000,
               18000, 120000, 95000, 380000],
    "stock": [12, 18, 45, 30, 50, 15, 20, 10]
}

# Crear un DataFrame
df_inventario = pd.DataFrame(inventario)

# Guardar los datos en un archivo Excel
df_inventario.to_excel("inventario.xlsx", index=False)

# Mostrar los datos
df_inventario
```

### 6. Importación y promedio de stock

## Ejercicio 2 - Inspección y visualización de datos con Matplotlib

En este ejercicio se utilizará el conjunto de datos `Automobile.csv` para realizar una inspección y visualización inicial de las características de los automóviles mediante Pandas y Matplotlib.

### 1. Carga e inspección inicial del conjunto de datos

Se carga el conjunto de datos de automóviles utilizando Pandas y se realiza una primera inspección de su contenido.

```python
# Cargar el conjunto de datos
df_automobile = pd.read_csv("Automobile.csv")

# Mostrar las primeras filas
df_automobile.head()
```

```python
# Información general del conjunto de datos
df_automobile.info()
```

### 2. Relación entre consumo de combustible y peso

Se utiliza un gráfico de dispersión para analizar la relación entre el peso de los automóviles (`weight`) y su consumo de combustible medido en millas por galón (`mpg`).

```python
import matplotlib.pyplot as plt

# Crear gráfico de dispersión
plt.scatter(df_automobile["weight"], df_automobile["mpg"])

# Agregar título y etiquetas
plt.title("Relación entre peso y consumo de combustible")
plt.xlabel("Peso del automóvil (weight)")
plt.ylabel("Millas por galón (mpg)")

# Agregar grilla
plt.grid()

# Mostrar el gráfico
plt.show()
```

**Interpretación:**  
El gráfico muestra una relación inversa entre el peso del automóvil y las millas por galón (`mpg`). A medida que aumenta el peso, el valor de `mpg` tiende a disminuir, lo que indica que los vehículos más pesados presentan, en general, un mayor consumo de combustible.

### 3. Distribución del consumo de combustible

Se utiliza un histograma para analizar la distribución del consumo de combustible de los automóviles, expresado en millas por galón (`mpg`).

```python
# Crear histograma del consumo de combustible
plt.hist(df_automobile["mpg"], bins=10, edgecolor="black")

# Agregar título y etiquetas
plt.title("Distribución del consumo de combustible")
plt.xlabel("Millas por galón (mpg)")
plt.ylabel("Cantidad de automóviles")

# Agregar grilla
plt.grid(axis="y")

# Mostrar el gráfico
plt.show()
```

**Interpretación:**  
El histograma muestra que la mayor cantidad de automóviles se concentra aproximadamente entre 15 y 30 millas por galón (`mpg`). A medida que aumenta el valor de `mpg`, la frecuencia disminuye, observándose pocos vehículos con valores superiores a 40 mpg. La distribución presenta una asimetría hacia la derecha.

### 4. Identificación de posibles valores atípicos

Se utilizan diagramas de caja (boxplots) para analizar las características `cylinders` y `origin` e identificar posibles valores atípicos. Debido a que la variable `origin` se encuentra expresada como texto en el conjunto de datos, se genera una representación numérica para su visualización.

```python
# Convertir temporalmente la variable origin a valores numéricos
origin_numerico = df_automobile["origin"].map({
    "usa": 1,
    "europe": 2,
    "japan": 3
})

# Crear el boxplot
plt.boxplot(
    [df_automobile["cylinders"], origin_numerico],
    tick_labels=["Cylinders", "Origin"]
)

# Agregar título y etiqueta
plt.title("Distribución de cilindros y origen")
plt.ylabel("Valores")

# Agregar grilla
plt.grid(axis="y")

# Mostrar el gráfico
plt.show()
```

**Interpretación:**  
Los diagramas de caja no muestran valores atípicos evidentes en las variables analizadas. En `cylinders`, los valores se encuentran entre 3 y 8 cilindros, con una mayor concentración en los valores inferiores. Para `origin`, se utilizó la codificación 1 = USA, 2 = Europa y 3 = Japón únicamente con fines de visualización.

### 5. Distribución de automóviles según año y origen

Se utiliza un gráfico de barras para analizar la cantidad de automóviles según el año del modelo (`model_year`) y su origen (`origin`).

```python
# Contar automóviles por año y origen
distribucion = pd.crosstab(
    df_automobile["model_year"],
    df_automobile["origin"]
)

# Crear gráfico de barras
distribucion.plot(kind="bar")

# Agregar título y etiquetas
plt.title("Distribución de automóviles según año y origen")
plt.xlabel("Año del modelo (model_year)")
plt.ylabel("Cantidad de automóviles")

# Agregar leyenda
plt.legend(title="Origen")

# Mostrar el gráfico
plt.show()
```

**Interpretación:**  
El gráfico muestra un predominio de automóviles de origen estadounidense durante gran parte del período analizado, especialmente en los primeros años. Hacia los años 80 se observa una mayor participación de vehículos de origen japonés, mientras que los automóviles europeos mantienen una presencia generalmente menor y relativamente estable.
